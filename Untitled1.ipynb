```python
# Drop duplicate rows explicitly creating a copy to prevent SettingWithCopyWarning
df = df.drop_duplicates().copy()

print("Dataset shape:", df.shape)
```

```python
# Create IsAlone feature based on the existing 'Family size' column
df["IsAlone"] = (df["Family size"] == 1).astype(int)

print(df[["Family size", "IsAlone"]].head())
```

```python
# Select appropriate features from the Online Food Delivery Dataset
X = df[["Age", "Gender", "Marital Status", "Occupation", "Monthly Income", "Educational Qualifications", "Family size", "Customer Type", "IsAlone"]]

# Define the target variable ('Output' indicates order success)
y = df["Output"]

print("Features (X) head:")
print(X.head())

print("\nTarget (y) head:")
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])

print("Training class proportions:")
print(y_train.value_counts(normalize=True))

print("Testing class proportions:")
print(y_test.value_counts(normalize=True))
```

```python
numeric_features = [
    "Age", "Fare", "FamilySize", "IsAlone", "CabinKnown"
]

categorical_features = [
    "Pclass", "Sex", "Embarked"
]
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore",
                              sparse_output=False))
])
```

```python
#Combining
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])
```

```python
# Redefine the actual features present in your food delivery dataset
numeric_features = ["Age", "Family size"]
categorical_features = [
    "Gender",
    "Marital Status",
    "Occupation",
    "Monthly Income",
    "Educational Qualifications",
    "Customer Type",
    "IsAlone"
]

# Re-initialize the ColumnTransformer with the correct columns
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

# Fit and transform the food delivery training and test data
X_train_prepared = preprocessor.fit_transform(X_train)
X_test_prepared = preprocessor.transform(X_test)

print("Training shape:", X_train_prepared.shape)
print("Testing shape:", X_test_prepared.shape)
```

```python
print("Missing values in training:",
      np.isnan(X_train_prepared).sum())

print("Missing values in testing:",
      np.isnan(X_test_prepared).sum())

print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

import matplotlib.pyplot as plt
import pandas as pd
```

```python
logistic_model = LogisticRegression(max_iter=1000)

logistic_model.fit(X_train_prepared, y_train)

logistic_predictions = logistic_model.predict(X_test_prepared)

print("First 10 predictions:", logistic_predictions[:10])
print("First 10 actual values:", y_test.iloc[:10].tolist())
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train_prepared, y_train)

knn_predictions = knn_model.predict(X_test_prepared)

print("First 10 KNN predictions:", knn_predictions[:10])
```

```python

tree_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

tree_model.fit(X_train_prepared, y_train)

tree_predictions = tree_model.predict(X_test_prepared)

print("First 10 tree predictions:", tree_predictions[:10])
```

```python
def evaluate_model(name, y_true, y_pred):
    print("\nModel:", name)
    print("Accuracy:", accuracy_score(y_true, y_pred))
    # Explicitly define pos_label as 'Yes'
    print("F1-score:", f1_score(y_true, y_pred, pos_label='Yes', zero_division=0))

    print("\nClassification report:")
    print(classification_report(
        y_true,
        y_pred,
        target_names=["No", "Yes"],
        zero_division=0
    ))

    cm = confusion_matrix(y_true, y_pred, labels=["No", "Yes"])
    display = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["No", "Yes"]
    )
    display.plot()
    plt.title(name + " Confusion Matrix")
    plt.show()
```

```python
evaluate_model("Logistic Regression", y_test, logistic_predictions)
evaluate_model("KNN", y_test, knn_predictions)
evaluate_model("Decision Tree", y_test, tree_predictions)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error

# Using columns that exist in the active Online Food Delivery Dataset
predictive_features = ["Family size", "IsAlone"]

X_reg_train = X_train[predictive_features]
X_reg_test = X_test[predictive_features]

# We will predict 'Age' as our continuous target variable
y_reg_train = X_train["Age"]
y_reg_test = X_test["Age"]

age_predictor_model = make_pipeline(
    SimpleImputer(strategy="median"),
    LinearRegression()
)

age_predictor_model.fit(X_reg_train, y_reg_train)

age_predictions = age_predictor_model.predict(X_reg_test)

print("Actual Ages:", y_reg_test.head().tolist())
print("Predicted Ages:", age_predictions[:5].tolist())

print("Mean Absolute Error:",
      mean_absolute_error(y_reg_test, age_predictions))
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "KNN",
        "Decision Tree"
    ],
    "Accuracy": [
        accuracy_score(y_test, logistic_predictions),
        accuracy_score(y_test, knn_predictions),
        accuracy_score(y_test, tree_predictions)
    ],
    "F1-score": [
        f1_score(y_test, logistic_predictions, pos_label='Yes'),
        f1_score(y_test, knn_predictions, pos_label='Yes'),
        f1_score(y_test, tree_predictions, pos_label='Yes')
    ]
})

print(results.round(3))
```
